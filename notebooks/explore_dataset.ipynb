## Barplot delle variabili categoriche

```python
import matplotlib.pyplot as plt

# Variabili categoriche identificate
categorical_vars = ['VehBrand', 'VehGas', 'Region', 'Area']

# Creare subplot
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.ravel()

# Creare barplot per ogni variabile categorica usando seaborn
for idx, var in enumerate(categorical_vars):
    ax = axes[idx]
    sns.countplot(data=df_train, x=var, ax=ax, hue=var, palette='Blues', legend=False)
    ax.set_title(f'Distribuzione di {var}', fontsize=12, fontweight='bold')
    ax.set_xlabel(var)
    ax.set_ylabel('Frequenza')
    plt.setp(ax.xaxis.get_majorticklabels(), rotation=45, ha='right')
    ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## istogramma delle variabili continue

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Identifica le variabili continue (numeriche)
continuous_vars = ['VehPower', 'VehAge', 'DrivAge', 'Density', 'BonusMalus']

# Imposta lo stile e la paletta colori coerente con i grafici precedenti
sns.set_theme(style="whitegrid")
palette = sns.color_palette("viridis", as_cmap=False)

# Crea gli istogrammi per ogni variabile continua
for col in continuous_vars:
    plt.figure(figsize=(10, 6))
    sns.histplot(df_train[col], kde=True, color=palette[0], bins=30)
    plt.title(f'Distribuzione di {col}')
    plt.xlabel(col)
    plt.ylabel('Frequenza')
    plt.show()
```

```python
# Calcola le statistiche descrittive per continuous_vars
stats_descrittive = df_train[continuous_vars].describe()
print(stats_descrittive)
```
