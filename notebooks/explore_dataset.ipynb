## Barplot delle variabili categoriche

```python
import matplotlib.pyplot as plt

# Variabili categoriche identificate
categorical_vars = ['VehBrand', 'VehGas', 'Region', 'Area']

# Creare subplot
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.ravel()

# Creare barplot per ogni variabile categorica usando seaborn
for idx, var in enumerate(categorical_vars):
    ax = axes[idx]
    sns.countplot(data=df_train, x=var, ax=ax, hue=var, palette='Blues', legend=False)
    ax.set_title(f'Distribuzione di {var}', fontsize=12, fontweight='bold')
    ax.set_xlabel(var)
    ax.set_ylabel('Frequenza')
    plt.setp(ax.xaxis.get_majorticklabels(), rotation=45, ha='right')
    ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## istogramma delle variabili continue

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Identifica le variabili continue (numeriche)
continuous_vars = ['VehPower', 'VehAge', 'DrivAge', 'Density', 'BonusMalus']

# Imposta lo stile e la paletta colori coerente con i grafici precedenti
sns.set_theme(style="whitegrid")
palette = sns.color_palette("viridis", as_cmap=False)

# Crea gli istogrammi per ogni variabile continua
for col in continuous_vars:
    plt.figure(figsize=(10, 6))
    sns.histplot(df_train[col], kde=False, color=palette[0], bins=30)
    plt.title(f'Distribuzione di {col}')
    plt.xlabel(col)
    plt.ylabel('Frequenza')
    plt.show()
```

```python
# Calcola le statistiche descrittive per continuous_vars
stats_descrittive = df_train[continuous_vars].describe()
print(stats_descrittive)
```

### barplot frequenza

```python
n_claims = df_train['ClaimNb'].nunique()
dark_blues = sns.color_palette("Blues", n_colors=n_claims + 3)[3:]  # evita le tonalità troppo chiare

plt.figure(figsize=(10, 6))
sns.countplot(data=df_train, x='ClaimNb', hue='ClaimNb', palette=dark_blues, legend=False)
plt.title('Distribuzione di ClaimNb')
plt.xlabel('ClaimNb')
plt.ylabel('Frequenza')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### distribuzione del costo

```python

# Seleziona solo i record con almeno un claim
df_claims = df_train[df_train['ClaimNb'] > 0].copy()

# Per scala logaritmica, escludo eventuali costi nulli/non positivi
df_claims_cost = df_claims[df_claims['claims_cost'] > 0]

plt.figure(figsize=(10, 6))
sns.histplot(data=df_claims_cost, x='claims_cost', bins=50, kde=True, color=palette[0])
#plt.xscale('log')
plt.title('Distribuzione del costo dei claim (ClaimNb > 0, asse X )')
plt.xlabel('claims_cost (scala normale)')
plt.ylabel('Frequenza')
plt.grid(axis='y', alpha=0.3)
plt.show()
```
